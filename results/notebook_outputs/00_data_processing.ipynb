## Cível - T2

### U1

```python
dataName = data.civel1()
df = pd.read_csv(f"{UTIL_PATH}/data/{dataName}.csv")
```

```python
df['dataInicio'] = pd.to_datetime(df['dataInicio'])
df = df.sort_values(by=['processoID', 'dataInicio'])
```

```python
def filtrar_fluxo(group):

    primeira_atividade = str(group.iloc[0]['activity'])
    ultima_atividade = str(group.iloc[-1]['activity'])

    comecou_distribuicao = 'Distribuição' in primeira_atividade
    terminou_definitivo = 'Definitivo' in ultima_atividade

    return comecou_distribuicao and terminou_definitivo

# 4. Aplicar o filtro por grupo de processoID
# Isso manterá todas as linhas dos processos que atendem ao critério
df_filtrado = df.groupby('processoID').filter(filtrar_fluxo)

# 5. Salvar o resultado em um novo CSV
df_filtrado.to_csv(f'{UTIL_PATH}/data/{dataName}_cleaned.csv', index=False)

# Exibir resultados básicos
print(f"Processos encontrados: {df_filtrado['processoID'].nunique()}")
print(f"O arquivo '{dataName}_cleaned.csv' foi gerado com sucesso.")
```
